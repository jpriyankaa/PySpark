```python
# Creating a spark session
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("MyApp").getOrCreate()
```

```python
# Data Cleaning
# drop nulls
df = df.dropna()
# Fill nulls in 'Salary' column
df = df.fillna({"Salary" : 0})
```

```python
# Drop duplicates
df = df.dropDuplicates()
```

```python
# checking cleaned data
df.show()
```

```python
# Renaming column name
df = df.withColumnRenamed("EmployeeID","EmpID")
```

```python
df.show()
```

```python
# Aggregations
df.groupBy("Department").count().show()
```

```python
from pyspark.sql.functions import avg, sum, min, max, col

window = Window.partitionBy("Department").orderBy(col("Salary").desc())
df = df.withColumn ("avg_salary",avg("Salary").over(window).cast("long"))
df = df.withColumn ("Total_salary",sum("Salary").over(window).cast("long"))
df = df.withColumn ("min_salary",min("Salary").over(window).cast("long"))
df = df.withColumn ("max_salary",max("Salary").over(window).cast("long"))
```

```python
# accessing specific columns from a table
df.select("avg_salary","Total_salary","min_salary","max_salary").show(5)
```

```python
df.groupBy("Department").agg({"Salary" : "Avg"}).show()
```

```python
from pyspark.sql.functions import avg, sum 
df.groupBy("Department").agg(avg("Salary").alias("AvgSalary")).show()
```

```python
# Featuring Engineering
# floor is used to remove decimal points
from pyspark.sql.functions import col,when, floor
df = df.withColumn("IsHigh",floor(when(col("Salary")>50000,1).otherwise(0)))
df = df.withColumn("Bonus", floor(col("Salary")*0.1))
```

```python
# using cast to remove decimals in bonus
from pyspark.sql.functions import col
df = df.withColumn("Bonus", (col("Salary")*0.1).cast("int"))
```

```python
df.show(10)
```

```python
# cast(Long) helps when you have long numbers and removes decimals
df = df.withColumn("Bonus", (col("Salary")*0.1).cast("Long"))
```

```python
df.show(5)
```

```python
# window functions
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, rank, dense_rank

window = Window.partitionBy("Department").orderBy(col("Salary").desc())
df = df.withColumn ("Rank", rank() .over(window))  # gaps in ranking
df = df.withColumn ("Dense_Rank", dense_rank().over(window)) # no gaps
df = df.withColumn ("Row_Number", row_number().over(window)) # unique row number
```

```python
df.show(10)
```

```python
df.groupBy("Dept").pivot("Year").agg(avg("Salary")).show()
```

```python
from pyspark.sql.functions import skewness

df.select(skewness("Salary")).show()
df.corr("Age", "Salary")
```

```python
# Checking Skewness
from pyspark.sql.functions import skewness

df.select(skewness("Salary")).show()
```

```python
# checking for correlation
df.corr("Age","Salary")
```

```python
from pyspark.sql.functions import first, last

df = df.withColumn("first_salary", first("Salary").over(window))
df = df.withColumn("last_salary", last("Salary").over(window))
```

```python
# UDF (User-Defined Function)
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

def category(salary):   # Create custom logic functions
    return "High" if salary > 50000 else "Low"

salary_udf = udf(category, StringType())
df = df.withColumn("SalaryLevel", salary_udf(df.Salary))
```

```python
# outliers -IQR filtering
q1, q3 = df.approxQuantile("Salary", [0.25, 0.75], 0.05)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

df = df.filter((df.Salary >= lower) & (df.Salary <= upper))
```

```python
# Date functions
from pyspark.sql.functions import year, month

df = df.withColumn("JoinYear", year(col("JoiningDate")))
```
